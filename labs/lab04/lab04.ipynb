### 4.2.5. Факторизация. Специальные матричные структуры

Рассмотрим несколько примеров. Для работы со специальными матричными структу-
рами потребуется пакет LinearAlgebra.
Решение систем линейный алгебраических уравнений 𝐴𝑥 = 𝑏:

```julia
# Задаём квадратную матрицу 3x3 со случайными значениями:
A = rand(3, 3)
```

```julia
# Задаём единичный вектор:
x = fill(1.0, 3)
```

```julia
# Задаём вектор b:
b = A*x
```

```julia
# Решение исходного уравнения получаем с помощью функции \
# (убеждаемся, что x - единичный вектор):
A\b
```

Julia позволяет вычислять LU-факторизацию и определяет составной тип факторизации
для его хранения:

```julia
# LU-факторизация:
Alu = lu(A)
```

Различные части факторизации могут быть извлечены путём доступа к их специаль-
ным свойствам:

```julia
# Матрица перестановок:
Alu.P
```

```julia
# Вектор перестановок:
Alu.p
```

```julia
# Матрица L:
Alu.L
```

```julia
# Матрица U:
Alu.U
```

Исходная система уравнений 𝐴𝑥 = 𝑏 может быть решена или с использованием
исходной матрицы, или с использованием объекта факторизации:

```julia
# Решение СЛАУ через матрицу A:
A\b
```

```julia
# Решение СЛАУ через объект факторизации:
Alu\b
```

Аналогично можно найти детерминант матрицы:

```julia
# Детерминант матрицы A:
det(A)
```

```julia
# Детерминант матрицы A через объект факторизации:
det(Alu)
```

Julia позволяет вычислять QR-факторизацию и определяет составной тип факториза-
ции для его хранения:

```julia
# QR-факторизация:
Aqr = qr(A)
```

По аналогии с LU-факторизацией различные части QR-факторизации могут быть из-
влечены путём доступа к их специальным свойствам:

```julia
# Матрица Q:
Aqr.Q
```

```julia
# Матрица R:
Aqr.R
```

```julia
# Проверка, что матрица Q - ортогональная:
Aqr.Q'*Aqr.Q
```

Далее рассмотрим примеры работы с матрицами большой размерности и специальной
структуры.

```julia
# Матрица 1000 х 1000:
n = 1000
A = randn(n,n)
```

```julia
# Симметризация матрицы:
Asym = A + A'
```

```julia
# Проверка, является ли матрица симметричной:
issymmetric(Asym)
```

Пример добавления шума в симметричную матрицу (матрица уже не будет симмет-
ричной):

```julia
# Добавление шума:
Asym_noisy = copy(Asym)
Asym_noisy[1,2] += 5eps()
```

```julia
# Проверка, является ли матрица симметричной:
issymmetric(Asym_noisy)
```

В Julia можно объявить структуру матрица явно, например, используя Diagonal,
Triangular, Symmetric, Hermitian, Tridiagonal и SymTridiagonal:

```julia
# Явно указываем, что матрица является симметричной:
Asym_explicit = Symmetric(Asym_noisy)
```

Далее для оценки эффективности выполнения операций над матрицами большой
размерности и специальной структуры воспользуемся пакетом BenchmarkTools:

```julia
import Pkg
Pkg.add("BenchmarkTools")
using BenchmarkTools
# Оценка эффективности выполнения операции по нахождению
# собственных значений симметризованной матрицы:
@btime eigvals(Asym);
```

```julia
# Оценка эффективности выполнения операции по нахождению
# собственных значений зашумлённой матрицы:
@btime eigvals(Asym_noisy);
```

```julia
# Оценка эффективности выполнения операции по нахождению
# собственных значений зашумлённой матрицы,
# для которой явно указано, что она симметричная:
@btime eigvals(Asym_explicit);
```

### 4.2.6. Общая линейная алгебра

Обычный способ добавить поддержку числовой линейной алгебры - это обернуть
подпрограммы BLAS и LAPACK. Собственно, для матриц с элементами Float32,Float64,
Complex {Float32} или Complex {Float64} разработчики Julia использовали такое же
решение. Однако Julia также поддерживает общую линейную алгебру, что позволяет,
например, работать с матрицами и векторами рациональных чисел.
Для задания рационального числа используется двойная косая черта:
1//2
В следующем примере показано, как можно решить систему линейных уравнений с ра-
циональными элементами без преобразования в типы элементов с плавающей запятой
(для избежания проблемы с переполнением используем BigInt):

```julia
# Матрица с рациональными элементами:
Arational = Matrix{Rational{BigInt}}(rand(1:10, 3, 3))/10
# Единичный вектор:
x = fill(1, 3)
# Задаём вектор b:
b = Arational*x
```

```julia
# Решение исходного уравнения получаем с помощью функции \
# (убеждаемся, что x - единичный вектор):
Arational\b
```

```julia
# LU-разложение:
lu(Arational)
```

### 4.4.4. Линейные модели экономики

1. Матрица 𝐴 называется продуктивной, если решение 𝑥 системы при любой неотрицательной правой части 𝑦 имеет только неотрицательные элементы 𝑥𝑖. Используя это определение, проверьте, являются ли матрицы продуктивными.

```julia
a = [1 2; 3 4]
```

```julia
b = 1/2*a
```

```julia
c = 1/10*a     
```

```julia
p_A = maximum(abs.(eigvals(a)))
println("A продуктивна? ", p_A < 1)
```

```julia
p_B = maximum(abs.(eigvals(b)))
println("B продуктивна? ", p_B < 1)
```

```julia
p_C = maximum(abs.(eigvals(c)))
println("C продуктивна? ", p_C < 1)
```

2. Критерий продуктивности: матрица 𝐴 является продуктивной тогда и только тогда, когда все элементы матрица (𝐸 − 𝐴)−1 являются неотрицательными числами. Используя этот критерий, проверьте, являются ли матрицы продуктивными.

```julia
a = [1 2; 3 1]
```

```julia
b = 1/2*a
```

```julia
c = 1/10*a
```

```julia
E = Matrix(I, 2, 2)
```

```julia
if all(inv(E - a) .>= 0)
    display(inv(E - a))
    println("Матрица A продуктивна")
else
    display(inv(E - a))
    println("Матрица A не продуктивна")
end
```

```julia
if all(inv(E - b) .>= 0)
    display(inv(E - b))
    println("Матрица B продуктивна")
else
    display(inv(E - b))
    println("Матрица B не продуктивна")
end 
```

```julia
if all(inv(E - c) .>= 0)
    display(inv(E - c))
    println("Матрица C продуктивна")
else
    display(inv(E - c))
    println("Матрица C не продуктивна")
end    
```

3. Спектральный критерий продуктивности: матрица 𝐴 является продуктивной тогда и только тогда, когда все её собственные значения по модулю меньше 1. Используя этот критерий, проверьте, являются ли матрицы продуктивными.

```julia
a = [1 2; 3 1]
```

```julia
b = 1/2*a
```

```julia
c = 1/10*a
```

```julia
d = [0.1 0.2 0.3; 0 0.1 0.2; 0 0.1 0.3]
```

```julia
aeig = eigen(a)
if all(abs.(aeig.values) .< 1)
    display(abs.(aeig.values))
    println("Матрица A продуктивна")
else
    display(abs.(aeig.values))
    println("Матрица A не продуктивна")
end
```

```julia
beig = eigen(b)
if all(abs.(beig.values) .< 1)
    display(abs.(beig.values))
    println("Матрица B продуктивна")
else
    display(abs.(beig.values))
    println("Матрица B не продуктивна")
end
```

```julia
ceig = eigen(c)
if all(abs.(ceig.values) .< 1)
    display(abs.(ceig.values))
    println("Матрица C продуктивна")
else
    display(abs.(ceig.values))
    println("Матрица C не продуктивна")
end
```

```julia
deig = eigen(d)
if all(abs.(deig.values) .< 1)
    display(abs.(deig.values))
    println("Матрица D продуктивна")
else
    display(abs.(deig.values))
    println("Матрица D не продуктивна")
end
```
